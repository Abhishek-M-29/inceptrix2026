```python
import git
import os
import shutil
import stat

# Function to handle read-only files on Windows during deletion
def remove_readonly(func, path, excinfo):
    os.chmod(path, stat.S_IWRITE)
    func(path)

class CloneProgress(git.RemoteProgress):
    def update(self, op_code, cur_count, max_count=None, message=''):
        print(f"Cloning: {message} ({cur_count}/{max_count or '?'})", end='\r')

repo_dir = "repos"
git_repo = "https://github.com/sankalp-happy/help-desk"
repo_name = git_repo.split("/")[-1].replace(".git", "")
target_path = os.path.join(repo_dir, repo_name)

try:
    if os.path.exists(target_path):
        print(f"Removing existing directory: {target_path}")
        shutil.rmtree(target_path, onerror=remove_readonly)
    
    os.makedirs(repo_dir, exist_ok=True)
    
    print(f"Cloning {git_repo}...")
    git.Repo.clone_from(git_repo, target_path, progress=CloneProgress())
    print(f"\nSuccessfully cloned into {target_path}")
    
except Exception as e:
    print(f"\nAn error occurred: {e}")
```
